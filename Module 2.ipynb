```python
#function for checking even or odd
def check_even_odd(num):
    if num%2==0:
        return "Even"
    else:
        return "Odd"
result=check_even_odd(5)
print("The number is ", result)
```

```python
#function for large  for two number
def find_largest(a, b):
    if a > b:
        return a
    else:
        return b
result = find_largest(10, 20)
print("The largest number is:", result)
```

```python
#function +loop how to print tables
def  multiplication_table(number):
    for i in range(1, 11):
        print(number, "x", i, "=", number*i)
number = int(input("Enter a number:"))
multiplication_table(number)
```

```python
#Largest of three Numbers
def largest(a, b, c):
    if a>=b and a>=c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c
a = int(input("Enter a:"))
b = int(input("Enter b:"))
c = int(input("Enter c:"))
print("Largest =", largest(a, b, c))
```

```python
# grade calculator
def find_grade(marks):
    if marks>=90:
        return"A"
    elif marks>=80:
        return"B"
    elif marks>=70:
        return"C"
    elif marks>=60:
        return"D" 
    elif marks>=40:
        return"E"
    else:
        return"FAIL"
marks = int(input("enter marks"))
print("Grade =", find_grade(marks))
```
